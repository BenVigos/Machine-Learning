```python
import keras
import pickle
from matplotlib import pyplot as plt
```

```python
model = keras.models.load_model("dqn_model_1.keras")
model.summary()

with open("training_data_1.pkl", "rb") as f:
    training_data = pickle.load(f)

# See what's inside
print(type(training_data))
print(training_data.keys() if isinstance(training_data, dict) else training_data)

```

```python

plt.plot(training_data['losses'])
plt.title("Training Loss")
plt.xlabel("Steps")
plt.ylabel("Loss")
plt.yscale("log")
plt.grid(True)
plt.show()

plt.plot(training_data['reward_history'])
plt.title("Reward Over Time")
plt.xlabel("Episodes")
plt.ylabel("Reward")
plt.grid(True)


plt.show()
```

```python
model_3 = keras.models.load_model("dqn_model_3.keras")
model_3.summary()

with open("training_data_3.pkl", "rb") as f:
    training_data_3 = pickle.load(f)

# See what's inside
print(type(training_data_3))
print(training_data_3.keys() if isinstance(training_data_3, dict) else training_data_3)
```

```python
import numpy as np 
# Loss plot
# Define the window size for the moving average
window_size = 50  # You can tweak this based on how smooth you want the curve

losses = training_data_3['losses']
x_loss = np.arange(len(losses))

# Compute moving average
if len(losses) >= window_size:
    moving_avg = np.convolve(losses, np.ones(window_size)/window_size, mode='valid')
    x_moving_avg = np.arange(window_size - 1, len(losses))
else:
    moving_avg = losses
    x_moving_avg = x_loss

# Plot
plt.plot(x_loss, losses, label="Raw Loss")
plt.plot(x_moving_avg, moving_avg, label=f"Moving Avg ({window_size})", color="orange")
plt.title("Training Loss")
plt.xlabel("Steps")
plt.ylabel("Loss")
plt.yscale("log")
plt.grid(True)
plt.legend()
plt.show()
rewards = training_data_3['reward_history']
x_reward = np.arange(len(rewards))

# Reward plot
plt.plot(x_reward, rewards, label="Reward")
if len(rewards) > 1:
    coeffs_reward = np.polyfit(x_reward, rewards, deg=1)
    fit_reward = np.poly1d(coeffs_reward)(x_reward)
    plt.plot(x_reward, fit_reward, "--", color="orange", label="Best Fit")

plt.title("Reward Over Time")
plt.xlabel("Episodes")
plt.ylabel("Reward")
plt.grid(True)
plt.legend()
plt.show()
print("Reward slope:", coeffs_reward[0])
```
