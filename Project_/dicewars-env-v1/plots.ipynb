```python
model_3 = keras.models.load_model("dqn_model_3.keras")
model_3.summary()

with open("training_data_3.pkl", "rb") as f:
    training_data_3 = pickle.load(f)

# See what's inside
print(type(training_data_3))
print(training_data_3.keys() if isinstance(training_data_3, dict) else training_data_3)
```

```python
import numpy as np 
# Loss plot
# Define the window size for the moving average
window_size = 50  # You can tweak this based on how smooth you want the curve

losses = training_data_3['losses']
x_loss = np.arange(len(losses))

# Compute moving average
if len(losses) >= window_size:
    moving_avg = np.convolve(losses, np.ones(window_size)/window_size, mode='valid')
    x_moving_avg = np.arange(window_size - 1, len(losses))
else:
    moving_avg = losses
    x_moving_avg = x_loss

# Plot
plt.plot(x_loss, losses, label="Raw Loss")
plt.plot(x_moving_avg, moving_avg, label=f"Moving Avg ({window_size})", color="orange")
plt.title("Training Loss")
plt.xlabel("Steps")
plt.ylabel("Loss")
plt.yscale("log")
plt.grid(True)
plt.legend()
plt.show()
rewards = training_data_3['reward_history']
x_reward = np.arange(len(rewards))

# Reward plot
plt.plot(x_reward, rewards, label="Reward")
if len(rewards) > 1:
    coeffs_reward = np.polyfit(x_reward, rewards, deg=1)
    fit_reward = np.poly1d(coeffs_reward)(x_reward)
    plt.plot(x_reward, fit_reward, "--", color="orange", label="Best Fit")

plt.title("Reward Over Time")
plt.xlabel("Episodes")
plt.ylabel("Reward")
plt.grid(True)
plt.legend()
plt.show()
print("Reward slope:", coeffs_reward[0])
```

```python
def test_agent(num_games, game=None, RENDER = False):
    wins = 0


    if game == None:
        game = Game(num_seats=4)

    for i in range(num_games):
        
        match = Match(game)

        grid, state = match.game.grid, match.state

        while True:
            currentplayer = players[state.player]

            action = currentplayer.get_attack_areas(grid, state)

            grid, state = match.step(action)

            if RENDER:
                match.render()

            if state.winner != -1 or state.player_num_dice[0] == 0:
                if state.winner == 0:
                    wins += 1
                break
                
        print(f"{wins}/{i+1}")
    return wins/num_games


GAMMA = 0.95
LEARNING_RATE = 0.01
MEMORY_SIZE = 10000
BATCH_SIZE = 32
EPISODES = 1000
EPSILON = 0  # Exploration factor
EPSILON_MIN = 0
EPSILON_DECAY = 0
TRAIN_AFTER_ACTIONS = 4
UPDATE_TARGET = 1000


agent = Player(MEMORY_SIZE=MEMORY_SIZE, EPSILON=EPSILON,
               LEARNING_RATE=LEARNING_RATE, BATCH_SIZE=BATCH_SIZE, GAMMA=GAMMA, EPSILON_MIN=EPSILON_MIN,
               EPSILON_DECAY=EPSILON_DECAY, model="dqn_model_5.keras")  # Our DQN player
players = [agent, RandomPlayer(), RandomPlayer(), RandomPlayer()]


stat = test_agent(num_games=100)
print(f"Rate: {stat}")
```
